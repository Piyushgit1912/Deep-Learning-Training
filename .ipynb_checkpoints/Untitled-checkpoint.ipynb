```python
import math
def sigmoid(x):
    return 1/(1+math.exp(-x))
sigmoid(100)
```

```python
import numpy as np 
import matplotlib.pyplot as plt

def plot_sigmoid():
    x=np.linspace(-10,10,100)
    y=1/(1+np.exp(-x))
    plt.plot(x,y)
    plt.xlabel('input')
    plt.ylabel('sigmoid output')
    plt.title('sigmoid function')
    plt.grid(True)
    plt.show()
plot_sigmoid()
```

## tanhyperbloic



```python
import math
def tanh(x):
    return (math.exp(x)-math.exp(-x))/(math.exp(x) + math.exp(-x))
tanh(50)
```

```python
import numpy as np 
import matplotlib.pyplot as plt
def plot_tanh():
    x=np.linspace(-10,10,100)
    y=np.tanh(x)
    plt.plot(x,y)
    plt.title('tanh activation function ')
    plt.grid(True)
    plt.show()
plot_tanh()
    
```

## threshol function 

```python
def threshold(x):
    return np.where(x>=0,1,0)
threshold(5)
```

## relu function 

```python
def relu(x):
    return max(0,x)
relu(-3)
relu(4)
```

```python
import numpy as np
import matplotlib.pyplot as plt
def plot_relu():
    x=np.linspace(-10,10,100)
    y=np.maximum(0,x)
    plt.plot(x,y)
    plt.grid(True)
    plt.show()
plot_relu()
```

## leaky relu 

```python
def leaky_relu(x):
    return max(0.1*x,x)
leaky_relu(-3)
```

```python
def plot_leaky_relu():
    x=np.linspace(-10,10,100)
    def leaky_relu(x,alpha=0.1):
        return np.where(x>=0,x,alpha*x)
    y=leaky_relu(x)
    plt.plot(x,y)
    plt.grid(True)
    plt.title('leaky relu')
    plt.show()
plot_leaky_relu()
```

## softmax

```python
import math 
def softmax(x):
    return math.exp(x)/np.sum(math.exp(x))
softmax(5)
```

```python
def plot_softmax():
    class_labels=['class a','class b','class c']
    probabillities=np.array([.2,.6,.3])
    plt.bar(class_labels,probabillities)
    plt.show()
plot_softmax()
```
