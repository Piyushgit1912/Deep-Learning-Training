# Logistic Regression & the XOR Problem

1. **Logistic regression** - one sigmoid neuron separates two linearly separable classes
2. **XOR** - not linearly separable, so a **hidden layer** is needed

Both use **binary cross-entropy**, the natural loss for sigmoid outputs.

```python
# Binary classification with logistic regression (one neuron + sigmoid).
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import SGD
import numpy as np

# Training data
# 6 samples with 2 features each
x_data = np.array([[1, 2],
                   [2, 3], 
                   [3, 1],
                   [4, 3],
                   [5, 3],
                   [6, 2]])
# class labels: 0 or 1
y_data = np.array([[0],
                   [0],
                   [0],
                   [1],
                   [1],
                   [1]])

# Build model
model = Sequential()
# 1 neuron + sigmoid = probability of class 1
model.add(Dense(1, input_shape=(2,), activation='sigmoid'))

# Compile with SGD optimizer
# plain stochastic gradient descent
sgd = SGD(learning_rate=0.1)
# binary cross-entropy matches a sigmoid output
model.compile(loss='binary_crossentropy', optimizer=sgd, metrics=['accuracy'])

# Model summary
model.summary()

# Train model (typo fixed: 'epochs' instead of 'eppochs')
model.fit(x_data, y_data, epochs=200, verbose=0)

# Predictions
# threshold the probability at 0.5 to get a class label
print("2,3 →", (model.predict(np.array([[2, 3]])) > 0.5).astype("int32"))
print("5,3 →", (model.predict(np.array([[5, 3]])) > 0.5).astype("int32"))
```
