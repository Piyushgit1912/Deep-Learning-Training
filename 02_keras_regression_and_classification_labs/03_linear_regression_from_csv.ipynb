# Linear Regression with Data Loaded from a CSV

Same model as the previous notebook (3 inputs -> 1 output), but the training data is loaded from
`data-01-test-score - data-01-test-score.csv` (student test scores: columns X1-X3 are the inputs, X4 is the target),
which must sit in the same folder as this notebook.

```python
# Load a CSV of numbers and train a linear-regression model on it.
from keras.models import Sequential
from keras.layers import Dense
import numpy as np

# load the CSV (lines starting with # are treated as comments): X1-X3 are the inputs, X4 is the target
xy = np.loadtxt("data-01-test-score - data-01-test-score.csv", delimiter=',')
# features = all columns except the last
x_data = xy[:, 0:-1]
# target = last column
y_data = xy[:, [-1]]

print("x_data", x_data)
print("y_data", y_data)

# a single linear neuron with 3 inputs
model = Sequential()
model.add(Dense(input_dim =3, units=1))
# MSE loss with the RMSprop optimiser
model.compile(loss='mse', optimizer='rmsprop')
# train silently (verbose=0) for 2000 epochs
model.fit(x_data, y_data, epochs=2000, verbose=0)

# predictions for two new input combinations
print("80, 75, 61", model.predict(np.array([[80, 75, 61]])))
print("0, 9, -1", model.predict(np.array([[0, 9, -1]])))
```
