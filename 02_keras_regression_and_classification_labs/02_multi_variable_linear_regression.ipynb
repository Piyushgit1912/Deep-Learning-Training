# Multi-Variable Linear Regression

Predict a **final score** from **three input scores** using a single linear neuron:

$$\hat{y} = w_1 x_1 + w_2 x_2 + w_3 x_3 + b$$

Trained with the RMSprop optimiser and the MSE loss.

```python
# Linear regression with 3 inputs in Keras.
from keras.models import Sequential
from keras.layers import Dense, Activation
from keras.optimizers import RMSprop
import numpy as np

# 5 samples x 3 input scores
x_data = np.array([[73.,80.,75.],
                  [93., 88., 93.],
                  [89., 91., 90.],
                  [96., 98., 100.],
                  [73., 66., 70.]])
# target value for each sample
y_data = np.array([[152.],
                   [185.],
                   [180.],
                   [196.],
                   [142.]])
print(x_data.shape)
print(y_data.shape)

model = Sequential()
# one neuron with 3 inputs and 1 output = linear regression
model.add(Dense(input_dim=3, units=1))
# 'linear' activation = identity (no non-linearity)
model.add(Activation('linear'))

# RMSprop adapts the learning rate for each parameter
rmsprop = RMSprop(learning_rate=0.1)
# MSE loss. NOTE: 'accuracy' is not a meaningful metric for regression
model.compile(loss='mse', optimizer=rmsprop, metrics=['accuracy'])
model.fit(x_data, y_data, epochs=1000, verbose=0)
```
