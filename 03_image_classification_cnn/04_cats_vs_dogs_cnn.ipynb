# Cats vs Dogs: Binary Image Classification with a CNN

A CNN that decides whether an image shows a **cat or a dog**, using the `cats_vs_dogs` dataset from TensorFlow Datasets and an efficient `tf.data` pipeline (resize -> normalise -> shuffle -> batch).

**Requires:** `pip install tensorflow-datasets`

Uses a sigmoid output with binary cross-entropy, and **Dropout** to reduce overfitting. The final plot compares training and validation accuracy.
