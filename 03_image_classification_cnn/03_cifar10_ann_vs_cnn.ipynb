# CIFAR-10: ANN vs CNN

CIFAR-10 contains 60,000 small **colour** images (32x32x3) in 10 classes (50,000 train / 10,000 test).

1. A **fully connected ANN** that ignores spatial structure
2. A **CNN** with two convolution + pooling blocks

The CNN reaches better accuracy with far fewer parameters.

```python
# Imports: TensorFlow/Keras for the models, Matplotlib for plots, NumPy for arrays.
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt 
import numpy as np
```

```python
# Load CIFAR-10 (downloaded automatically on first use).
# load datasets 
(X_train, y_train), (X_test, y_test)= datasets.cifar10.load_data()
```

```python
# Test set shape: (10000, 32, 32, 3).
X_test.shape
```

```python
# Training set shape: (50000, 32, 32, 3).
X_train.shape
```

```python
# Labels arrive with shape (N, 1); flatten them to (N,) so they can be used as simple indices.
y_train= y_train.reshape(-1,)
# -1 means single column conversion
```

```python
# Same flattening for the test labels.
y_test=y_test.reshape(-1,)
```

```python
# Class names.
# NOTE: this list has a typo ('dear' should be 'deer') and the order differs from CIFAR-10's official label order:
# airplane, automobile, bird, cat, deer, dog, frog, horse, ship, truck.
# As written, 'cat' and 'bird' are swapped, so plots/predictions for those two classes get the wrong name.
classes=['airplane','automobile','cat','bird','dear','dog','frog','horse','ship','truck']
```

```python
# Helper that shows one image together with its class name.
def plot_sample(X,y,index):
    plt.figure(figsize=(5,2))
    plt.imshow(X[index])
    plt.xlabel(classes[y[index]])
    
```

```python
# Show training image 0.
plot_sample(X_train,y_train,0)
```

```python
# Show training image 1.
plot_sample(X_train,y_train,1)
```
