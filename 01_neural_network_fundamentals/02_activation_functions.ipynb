# Activation Functions

An activation function adds **non-linearity** to a neuron, which is what lets a network learn complex patterns. For each function below we implement it from scratch and then plot it.

| Function | Formula | Output range | Typical use |
|---|---|---|---|
| Sigmoid | 1 / (1 + e^-x) | (0, 1) | binary output layer |
| tanh | (e^x - e^-x) / (e^x + e^-x) | (-1, 1) | hidden layers (zero-centred) |
| Threshold (step) | 1 if x >= 0 else 0 | {0, 1} | historical perceptron |
| ReLU | max(0, x) | [0, inf) | default for hidden layers |
| Leaky ReLU | max(alpha*x, x) | (-inf, inf) | avoids "dying" ReLU units |
| Softmax | e^xi / sum(e^xj) | (0, 1), sums to 1 | multi-class output layer |

# Activation function

## Sigmoid

```python
# Sigmoid in pure Python. Large |x| saturates near 0 or 1, where gradients vanish.
import math
def sigmoid(x):
    return 1/(1+ math.exp(-x))
sigmoid(100)
```

```python
# Plot sigmoid over [-10, 10] - the classic S-shaped curve.
import numpy as np
import matplotlib.pyplot as plt

def plot_sigmoid():
    # 100 evenly spaced inputs between -10 and 10
    x = np.linspace(-10, 10, 100)
    y = 1/(1+np.exp(-x))
    plt.plot(x,y)
    plt.xlabel("Input")
    plt.ylabel("Sigmoid Output")
    plt.title("Sigmoid Activation Function")
    plt.grid(True)
    plt.show()
plot_sigmoid()
```

## tanh(tanhyperbloic)

```python
# tanh in pure Python. Like sigmoid but zero-centred, with outputs in (-1, 1).
import math

def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))
tanh(50)
```

```python
# Plot tanh using NumPy's built-in np.tanh.
import numpy as np
import matplotlib.pyplot as plt

def plot_tanh():
    x = np.linspace(-10, 10, 100)
    y = np.tanh(x)
    plt.plot(x,y)
    plt.xlabel("Input")
    plt.ylabel("tanh Output")
    plt.title("tanh Activation Function")
    plt.grid(True)
    plt.show()
plot_tanh()
```

## Threshold Functiom

```python
# Threshold (step) function: 1 if x >= 0 else 0. Not differentiable, so it cannot be trained with gradient descent.
import numpy as np
def threshold(x):
    return np.where( x>=0,1,0)
threshold(5)
```

```python
# Plot of the step function.
# NOTE: this plot switches at x >= 1, whereas the previous cell uses x >= 0.
import numpy as np
import matplotlib.pyplot as plt

def plot_threshold():
    x = np.linspace(-10, 10, 100)
    def threshold(x):
        return np.where(x>=1,1,0)
    y = threshold(x)
    plt.plot(x,y)
    plt.xlabel("Input")
    plt.ylabel("Threshold Output")
    plt.title("Threshold Activation Function")
    plt.grid(True)
    plt.show()
plot_threshold()
```

## ReLU

```python
# ReLU: max(0, x). Cheap to compute and does not saturate for positive inputs.
def relu(x):
    return max(0,x)
print(relu(-4))
print(relu(9))
```

```python
# Plot ReLU: flat at 0 for negative inputs, a straight line for positive inputs.
import numpy as np
import matplotlib.pyplot as plt

def plot_Relu():
    x = np.linspace(-10, 10, 100)
    y = np.maximum(0,x)
    plt.plot(x,y)
    plt.xlabel("Input")
    plt.ylabel("Relu Output")
    plt.title("Relu Activation Function")
    plt.grid(True)
    plt.show()
plot_Relu()
```
