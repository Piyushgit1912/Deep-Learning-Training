# Loss (Cost) Functions

A loss function measures **how wrong** the model's predictions are; training means minimising it.

**Regression losses:** MAE, MSE, Huber
**Classification losses:** Binary cross-entropy, Hinge, Categorical cross-entropy

Every function is implemented from scratch with NumPy and tested on a tiny example.

## Regression Cost Function
1. MAE (Mean Absolute Error)
2. MSE (Mean Square Error)
3. Hubber loss (Mean Square Error + MAE)

```python
# Mean Absolute Error: the average size of the errors. Robust to outliers; its gradient has constant magnitude.
# MAE = 1/n SUM(|y - y'|)
import numpy as np

def mean_absolute_error(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))

y_true = np.array([1, 2, 3, 4, 5])
y_pred = np.array([1.1, 2.2, 2.9, 4.1, 4.8])

mae = mean_absolute_error(y_true, y_pred)
print(f"MAE : {mae}")
```

```python
# Mean Squared Error: squaring penalises large errors much more than small ones (sensitive to outliers).
# MSE = 1/n SUM((y - y')^2)
import numpy as np

def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
y_true = np.array([1, 2, 3, 4, 5])
y_pred = np.array([1.1, 2.2, 2.9, 4.1, 4.8])
mse = mean_squared_error(y_true, y_pred)
print(f"MSE : {mse}")
```
