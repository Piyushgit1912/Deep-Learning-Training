# Single Neuron & Forward Pass

**Goal:** understand how a neural network computes its output, using only NumPy.

We build up in three steps:

1. **One neuron** with two inputs and a sigmoid activation
2. **One layer** of two neurons using matrix multiplication and ReLU
3. **A tiny full network** (hidden layer -> output neuron)

**Key idea:** every neuron computes a weighted sum of its inputs plus a bias, then passes it through an activation function:

$$z = \sum_i w_i x_i + b \qquad a = f(z)$$

# 1=INput 1=Neuron (Using Sigmoid)

```python
# One neuron with two inputs. z = x1*w1 + x2*w2 (bias omitted), then sigmoid squashes z into (0, 1).
import numpy as np

x1, x2 = 1.0, 0.5
w1, w2 = 0.3, 0.2

# Sigmoid: 1 / (1 + e^-x) maps any real number into the range (0, 1)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
# Weighted sum of the inputs (the pre-activation value)
z = (x1 * w1) + (x2 * w2)
# Apply the activation function to get the neuron's output
o = sigmoid(z)
print("Output:", o)
print("Weight Sum",z)
```

# 1=input 2=Neuron (Using ReLU)

```python
# A layer of 2 neurons fed by the same 2 inputs, computed with matrix multiplication.
# Each ROW of W holds the weights of one neuron, so Z = X . W^T + B gives both pre-activations at once.
import numpy as np

# Input vector (2 features)
X = np.array([1.0, 0.5])

# Weight matrix, shape (neurons, inputs)
W = np.array([[0.2, 0.4],  
              [0.6, 0.9]])  
# One bias per neuron
B = np.array([0.1, 0.2])

# ReLU: max(0, x) - passes positive values, zeroes out negatives
def relu(x):
    return np.maximum(0, x)
# Weighted sum for the whole layer in one step
Z = np.dot(X, W.T) + B
# Layer output after the activation
O = relu(Z)
print("Output:", O)
```

# Ful Neural Network

```python
# Complete forward pass: input -> hidden layer (ReLU) -> output neuron (sigmoid).
import numpy as np

X = np.array([1.0, 0.5])
W = np.array([[0.2, 0.4],
              [0.6, 0.9]])
B = np.array([0.1, 0.2])
def relu(x):
    return np.maximum(0, x)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
# Hidden layer: weighted sum + ReLU
Z = np.dot(X, W.T) + B
O = relu(Z)
# Output-neuron weights and bias (hidden -> output)
w5, w6 = 0.4, 0.5
b3 = 0.2
# Output neuron: weighted sum of the hidden activations
z3 = (O[0] * w5) + (O[1] * w6) + b3
# Sigmoid output in (0, 1) - can be read as a probability
output = sigmoid(z3)

print("Hidden Layer Output:", O)
print("Final Output (Sigmoid):", output)
```
