# Categorical Cross-Entropy (scratch notes)

A short scratch notebook with the categorical cross-entropy loss. See `03_loss_functions.ipynb` for the full set of loss functions.

```python
# Categorical cross-entropy: -(1/n) * sum(y_true * log(y_pred)).
import numpy as np 
def categorical_cross(y_true,y_pred):
    # number of samples, used to average the loss
    n_samples=y_true.shape[0]
    # tiny constant to avoid log(0)
    epsilon=1e-15
    # keep predictions inside (epsilon, 1 - epsilon)
    y_pred=np.clip(y_pred,epsilon,1-epsilon)
    loss=-np.sum(y_true*np.log(y_pred))/n_samples
    return loss
y_true=np.array([1,0,1,0,1])
y_pred=np.array([0.9,0.1,.6,.3,.8])
categorical_loss=categorical_cross(y_true,y_pred)
print(categorical_loss, 'donee')
```
