# Optimizers

Variations of gradient descent, implemented from scratch on the same toy problem: learn `y = 2x + 3` (parameters `w` and `b`).

| Optimizer | Idea |
|---|---|
| Mini-batch gradient descent | update using small batches instead of the whole dataset |
| SGD with momentum | keep a velocity so updates build speed in consistent directions |
| Nesterov accelerated gradient | compute the gradient at the 'look-ahead' position |
| Adagrad | per-parameter learning rate that shrinks as squared gradients accumulate |
| Adadelta | *(placeholder - not implemented yet)* |

```python
# Mini-batch gradient descent: each update uses only `batch_size` samples.
# NOTE: the loss is recorded after every mini-batch update (not once per epoch as the inline comment below says).
# mini Batch Gradient
import numpy as np
# data
X = np.array([1, 2, 3, 4, 5], dtype = np.float64)
Y = 2* X+3

# Initialize Parameter
# model: y = w*x + b, both parameters start at 0
w = 0
b = 0
lr = 0.01
# number of samples used per parameter update
batch_size = 2
losses = []

# loop
for epoch in range(300):
    # walk through the data in chunks of `batch_size`
    for i in range(0, len(X), batch_size):
        # Batch create
        xb = X[i:i+batch_size]
        yb = Y[i:i+batch_size]

        # prediction
        pred = w*xb+b

        # derivation of lossfunction(Mean Square error)
        dw = (-2/len(xb))*np.sum(xb*(yb-pred))
        db = (-2/len(xb))*np.sum(yb-pred)

        # Weight update
        w -=lr*dw
        b -=lr*db

        # Loss calculation for each epoch
        full_pred = w*X+b
        loss = np.mean((Y-full_pred)**2)
        losses.append(loss)

print("Optimal weight:", w)
print("Optimal bias:", b)
predicted = w*X+b
```

```python
# Momentum: v <- beta*v + lr*gradient ; w <- w - v.
# The velocity is an exponentially decaying average of past gradients, which speeds up learning and damps oscillations.
# SGD With Momentum
import numpy as np

X = np.array([1, 2, 3, 4, 5], dtype=float)
Y = np.array([5, 7, 9, 11, 13], dtype=float)

# Initialize parameters
w, b = 0, 0
# velocity terms start at zero
vw, vb = 0, 0
beta = 0.9   # momentum (usually closer to 0.9, not 0.09)
lr = 0.01

# Training loop
for epoch in range(300):
    pred = w * X + b
    dw = (-2 / len(X)) * np.sum(X * (Y - pred))
    db = (-2 / len(X)) * np.sum(Y - pred)

    # Momentum updates
    vw = beta * vw + lr * dw
    vb = beta * vb + lr * db
    w -= vw
    b -= vb

print("Final weights:", w)
print("Final bias:", b)
```

```python
# Nesterov accelerated gradient: the gradient is evaluated at the look-ahead position (where momentum is about to take us) for a better correction.
# Nesterov Accelraalted Gradiect Method

import numpy as np

X = np.array([1, 2, 3, 4, 5], dtype = float)
Y = np.array([5, 7, 9, 11, 13], dtype = float)
# Initialize parameters
w, b = 0, 0
vw, vb = 0, 0
beta = 0.9   
lr = 0.01

# Loop
for epoch in range(300):
    # look-ahead position
    look_w = w-beta*vw
    look_b = b-beta*vb
    # pred
    pred = look_w* X + look_b
    dw = (-2 / len(X)) * np.sum(X * (Y - pred))
    db = (-2 / len(X)) * np.sum(Y - pred)

     # Momentum updates
    vw = beta * vw + lr * dw
    vb = beta * vb + lr * db
    w -= vw
    b -= vb
print("Final weights:", w)
print("Final bias:", b)
```
